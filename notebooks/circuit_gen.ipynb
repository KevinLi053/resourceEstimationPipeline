```python
def cube_plot(t_vals, q_vals):
    import matplotlib.pyplot as plt
    T, Q = np.meshgrid(t_vals, q_vals)
    plt.figure(figsize=(6, 6))
    plt.scatter(T.ravel(), Q.ravel(), s=40)
    plt.xlabel("t")
    plt.ylabel("q")
    plt.title("Grid of (t, q) combinations")
    plt.grid(True)
    plt.show()
```

```python
def circuit_cube(n_circuits: int, t_range: list[int], q_range: list[int]):
    from qiskit import QuantumCircuit
    import numpy as np

    t_vals = np.round(np.linspace(t_range[0], t_range[1], n_circuits)).astype(int)
    q_vals = np.round(np.linspace(q_range[0], q_range[1], n_circuits)).astype(int)

    #cube_plot(t_vals, q_vals)

    circuits = []

    for t in t_vals:
        for q in q_vals:
            qc = QuantumCircuit(q, 0)
            for i in range(t):
                current = i % q
                nxt = (i + 1) % q
                qc.h(current)
                qc.t(current)
                qc.cx(current, nxt)
            circuits.append(qc)
    return circuits
```

```python
from qiskit.qasm2 import dumps
def export_to_qasm(circuits):
    QASM_circuits = []
    for circuit in circuits:
        qasm_source = dumps(circuit)
        QASM_circuits.append(qasm_source)
    return QASM_circuits
```

```python
c_count = 50
t_count = 5000
q_count = 1000

circuits = circuit_cube(c_count, [1, t_count], [2, q_count])
print("Circuit Count: " + str(len(circuits)))
```

```python
qasm_circuits = export_to_qasm(circuits)
print("QASM Circuit Count: " + str(len(circuits)))
```

```python
# save circuits as qasm files sorted as t1(all q), t2(all q), t3(all q) ... t5000(all q)

import os
from qiskit.qasm3 import dumps

os.makedirs("qasm3_circuits", exist_ok=True)

for i, circuit in enumerate(circuits):
    qasm3_source = dumps(circuit)
    with open(f"qasm3_circuits/circuit_{i}.qasm3", "w") as f:
        f.write(qasm3_source)
```
