```python
import numpy as np

def cube_plot(t_vals, q_vals):
    import matplotlib.pyplot as plt
    T, Q = np.meshgrid(t_vals, q_vals)
    plt.figure(figsize=(6, 6))
    plt.scatter(T.ravel(), Q.ravel(), s=40)
    plt.xlabel("t")
    plt.ylabel("q")
    plt.title("Grid of (t, q) combinations")
    plt.grid(True)
    plt.show()
```

```python
def circuit_cube(n_circuits: int, t_range: list[int], q_range: list[int]):
    from qiskit import QuantumCircuit

    t_vals = np.round(np.linspace(t_range[0], t_range[1], n_circuits)).astype(int)
    q_vals = np.round(np.linspace(q_range[0], q_range[1], n_circuits)).astype(int)

    #cube_plot(t_vals, q_vals)

    circuits = {}

    for t in t_vals:
        for q in q_vals:
            qc = QuantumCircuit(q, 0)
            for i in range(t):
                current = i % q
                nxt = (i + 1) % q
                qc.h(current)
                qc.t(current)
                qc.cx(current, nxt)
            circuits[f"{q}_{t}"] = (qc)
    return circuits


# ---------------------------------------------------------------------------
# Helper: synthesize synthetic circuits for benchmarking
# ---------------------------------------------------------------------------

def t_circuit_distribution(n_circuits: int, t_max: int, q_max: int, linear: bool = True):
    """Return a list of synthetic Clifford+T circuits.

    Each circuit has *q* qubits and approximately *t* T gates distributed
    linearly or geometrically across the range [1, t_max] x [2, q_max].

    Parameters
    ----------
    n_circuits : int  number of circuits to generate
    t_max      : int  max T gate count per circuit
    q_max      : int  max qubit count per circuit
    linear     : bool  if True use linspace; otherwise use geomspace

    Returns
    -------
    list[QuantumCircuit]
    """
    from qiskit import QuantumCircuit
    import numpy as np

    t_vals = np.linspace(1, t_max, n_circuits) if linear else np.geomspace(1, t_max + 1, n_circuits)
    q_vals = np.linspace(2, q_max, n_circuits) if linear else np.geomspace(2, q_max + 1, n_circuits)

    circuits = []
    for t, q in zip(t_vals, q_vals):
        qc = QuantumCircuit(int(q), 0)
        for i in range(int(t)):
            current = i % int(q)
            nxt = (i + 1) % int(q)
            qc.h(current)
            qc.t(current)
            qc.cx(current, nxt)
        circuits.append(qc)
    return circuits
```

```python
c_count = 15
t_count = 5000
q_count = 1000
cube = True

if cube:
    circuits = circuit_cube(c_count, [1, t_count], [2, q_count])
    print("Circuit Count: " + str(len(circuits)))
else:
    circuits = t_circuit_distribution(c_count, t_count, q_count, True)
    print("Circuit Count: " + str(len(circuits)))
```

```python
# save circuits as qasm files sorted as t1(all q), t2(all q), t3(all q) ... t5000(all q)

import os
from qiskit import qasm3

if cube:
    output_dir = os.path.join(
        "qasm3_circuits",
        "cube",
        f"{c_count * c_count}_{q_count}_{t_count}"
    )
else:
    output_dir = os.path.join(
        "qasm3_circuits",
        "linear",
        f"{c_count * c_count}_{q_count}_{t_count}"
    )
    
os.makedirs(output_dir, exist_ok=True)

for k, v in circuits.items():
    qasm_path = os.path.join(output_dir, f"{k}.qasm3")
    with open(qasm_path, "w") as f:
        qasm3.dump(v, f)
```
